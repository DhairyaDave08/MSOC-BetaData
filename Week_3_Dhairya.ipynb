## **Problem 1: Array Statistics**


Given a NumPy array, find:

* Sum of all elements
* Mean of all elements
* Maximum element

### Example

```python
arr = np.array([1, 2, 3, 4, 5])
```

### Expected Output

```text
Sum = 15
Mean = 3
Max = 5
```



````python
a=np.array([1,2,3,4,5])
print("Sum: ",np.sum(a))
print("Mean: ",np.mean(a))
print("Maximum: ",np.max(a))
````

## Problem 3: Row With Maximum Sum


Given a 2D NumPy array, find the index of the row having the largest sum.

### Example

```python
arr = np.array([
    [1, 2, 3],
    [5, 5, 5],
    [2, 2, 2]
])
```

### Expected Output

```text
1
```

### Explanation

```text
Row sums = [6, 15, 6]
```

````python
d=np.array([[1,2,3],[5,5,5],[2,2,2]])
e=np.sum(d,axis=1) # results into the sum of each row
print(np.argmax(e)) # gives the index having maximum value
````

## Problem 4: Student Performance Analysis


Each row represents a student and each column represents marks in a subject.

Return the indices of students whose average marks are greater than `75`.

### Example

```python
marks = np.array([
    [80, 90, 70],
    [50, 60, 70],
    [90, 95, 85]
])
```

### Expected Output

```python
[0, 2]
```

### Explanation

```text
Student 0 Average = 80
Student 1 Average = 60
Student 2 Average = 90
```


````python
marks = np.array([
    [80, 90, 70],
    [50, 60, 70],
    [90, 95, 85]
])
avg_m=np.mean(marks,axis=1)
print(np.where(avg_m>75)[0])
````

## Problem 5 Network Traffic Anomaly Detection (Very Hard Version)

A server records requests per minute.
<br>
You are given traffic data for many servers.
<br>
Return the indices of servers whose traffic is more than 2× the overall average traffic. <br>

```py
traffic = np.array([
 [100,120,110],
 [90,95,100],
 [500,550,600]
])
```
```py
[2]
```

````python
traffic = np.array([
 [100,120,110],
 [90,95,100],
 [500,550,600]
])
t_avg=np.mean(traffic)
i_avg=np.mean(traffic,axis=1)
print(np.where(i_avg>=2*t_avg)[0])
````

## Problem 6 Fraud Detection Feature Generator (Very Hard Version)
You are working for a fintech company. Every row represents a customer transaction.
<br>
Columns:
```txt
[Amount, TimeSpent, FailedAttempts]
```
<br>
Risk Score:

```
0.6 × normalized(Amount)
+
0.3 × normalized(TimeSpent)
+
0.1 × normalized(FailedAttempts)
```

Return top K most risky users.
<br>
Input :
```py
transactions =
[
[1000,50,0],
[5000,20,2],
[3000,80,5]
]

k = 2
```

````python
transactions = np.array([
    [1000, 50, 0],
    [5000, 20, 2],
    [3000, 80, 5]
])
k = 2
mn = transactions.min(axis=0)
mx = transactions.max(axis=0)
norm = (transactions - mn) / (mx - mn)
risk = (0.6 * norm[:, 0] +0.3 * norm[:, 1] +0.1 * norm[:, 2])
top_k = np.argsort(risk)[::-1][:k]
print(top_k)
````
